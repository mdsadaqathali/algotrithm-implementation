```python
from p import load_data,clean_data,preprocess_data
from t import train_model
from e import evaluate_model

data=load_data("Breast_Cancer[1].csv")
data=clean_data(data)
X_train,X_test,y_train,y_test=preprocess_data(data)
model=train_model(X_train,y_train)
acc,prec,rec,f1,roc,cm=evaluate_model(model,X_test,y_test)

print("Accuracy:",acc)
print("Precision:",prec)
print("Recall:",rec)
print("F1 Score:",f1)
print("ROC-AUC:",roc)
print("Confusion Matrix:\n",cm)
print("Total predictions:",len(y_test))
```
